```python
import pandas as pd
import dask.dataframe as dd
import psutil
import time
import tracemalloc  
import csv
import sys
```

```python
def chunking_memory_usage(file_path, chunksize=1000):
    tracemalloc.start()  
    start_time = time.time()

    for chunk in pd.read_csv(file_path, chunksize=chunksize):
        pass 
    chunking_time = time.time() - start_time
    mem_usage = tracemalloc.get_traced_memory()[1] / 1024**2  
    tracemalloc.stop()
    
    print(f"Chunking: Time = {chunking_time:.2f}s, Memory Usage = {mem_usage:.2f} MB")
    return chunking_time, mem_usage


chunking_time, chunking_mem = chunking_memory_usage("./emails.csv")
```

```python
def dask_memory_usage(file_path):
    
    csv.field_size_limit(sys.maxsize)

    tracemalloc.start()
    start_time = time.time()

    df_dask = dd.read_csv("emails.csv", engine="python", on_bad_lines="skip", quotechar='"')
    df_dask.head()  

    dask_time = time.time() - start_time
    mem_usage = tracemalloc.get_traced_memory()[1] / 1024**2
    tracemalloc.stop()
    
    print(f"Dask: Time = {dask_time:.2f}s, Memory Usage = {mem_usage:.2f} MB")
    return dask_time, mem_usage


dask_time, dask_mem = dask_memory_usage("./emails.csv")
```

```python
def compressed_memory_usage(file_path):
    tracemalloc.start()
    start_time = time.time()

    df_compressed = pd.read_csv(file_path, compression='gzip')

    compressed_time = time.time() - start_time
    mem_usage = tracemalloc.get_traced_memory()[1] / 1024**2
    tracemalloc.stop()
    
    print(f"Compressed: Time = {compressed_time:.2f}s, Memory Usage = {mem_usage:.2f} MB")
    return compressed_time, mem_usage


compressed_time, compressed_mem = compressed_memory_usage("./emails.csv.gz")
```

```python

comparison_df = pd.DataFrame({
    "Method": ["Chunking", "Dask", "Compressed"],
    "Time (s)": [chunking_time, dask_time, compressed_time],
    "Memory Usage (MB)": [chunking_mem, dask_mem, compressed_mem]
})

print(comparison_df)
```
